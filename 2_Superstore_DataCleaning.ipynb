**2.5. Fix Data Types**

```python
# Convert the date columns to pandas datatime datatype
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Check the datatype of the columns
df[['Order Date', 'Ship Date']].dtypes
```

```python
# Converting text columns to string data type and remove whitespaces
text_columns = ['Order ID', 'Ship Mode', 'Customer ID', 'Customer Name', 'Segment', 'Country', 'City', 'State','Region', 'Product ID', 'Category', 'Sub-Category',
       'Product Name']

for column in text_columns:
  df[column] = df[column].astype('string').str.strip()
```

```python
df.info()
```

**2.5. Handle duplicates in the dataset**

```python
# 1. Remove Duplicates
df = df.drop_duplicates()
```

2.5.1 Check and fix duplicated Order ID and Product ID pair

```python
# Number of duplicate orders
Duplicate_Order = df['Order ID'].duplicated().sum()
print(Duplicate_Order)

# Number of Unique Product ID for each Order ID
Test1=df.groupby('Order ID')['Product ID'].nunique().sort_values(ascending=False)
print(Test1.max())
```

```python
# Investigate duplicates orders
j=0
for i in range(0, Test1.max() + 1):
  print(f'The number of order id that has  {i} Product ID ',len(Test1[Test1==i]))
  j += i * len(Test1[Test1 == i])
print('Numbre of order id including duplicate with different product Id is ',j)
```

```python
# Filter and display all duplicate entries based on Order ID and Product ID
duplicates = df[df.duplicated(subset=['Order ID', 'Product ID'], keep=False)]

# Sort them so matching duplicates are right next to each other
duplicates_sorted = duplicates.sort_values(by=['Order ID', 'Product ID'])

# View key columns of the duplicate rows
duplicates_sorted[['Order ID', 'Order Date', 'Ship Date','Ship Mode','Product ID', 'Product Name', 'Segment', 'City','State', 'Postal Code','Region','Category','Sub-Category']]
```

```python
len(duplicates_sorted)
```

```python
# 1. Identify columns to update
target_cols = ['Sales', 'Quantity', 'Profit']
df['Quantity'] = df['Quantity'].astype(float)

# 2. Calculate group means for duplicate Order ID + Product ID pairs and assign directly
df.loc[duplicates_sorted.index, target_cols] = df.groupby(['Order ID', 'Product ID'])[target_cols].transform('mean')

# 3. Now drop the duplicate rows, keeping the first occurrence (which now holds the mean values)
df_cleaned = df.drop_duplicates(subset=['Order ID', 'Product ID','Product Name'], keep='first')

df = df_cleaned

df.shape
```

2.5.2 Check and fix duplicated Product ID and Product Name pair

```python
# Investigate data for consistency between Product ID and Product Name
print('the number of product name', len(df['Product Name'].unique()))
print('the number of product ID',len(df['Product ID'].unique()))
```

```python
# Check product IDs with more than one product name
Test=df.groupby('Product ID')['Product Name'].nunique().sort_values(ascending=False)
print(len(Test[Test==2]))
print(len(Test[Test==1]))
print(len(Test[Test==0]))
```

```python
# Display all product ID that have 2 products name  along with product name
pro_ids=Test[Test==2].index
df[df['Product ID'].isin(pro_ids)][['Product ID', 'Product Name']].drop_duplicates().sort_values('Product ID')
```

```python
# Calculate the most frequent name ONLY for the 32 affected Product IDs
most_frequent_names = (df[df['Product ID'].isin(pro_ids)].groupby('Product ID')['Product Name'].agg(lambda x: x.mode()[0]).to_dict())

# Convert dictionary items to a clean 2-column DataFrame
d = pd.DataFrame(list(most_frequent_names.items()), columns=['Product ID', 'Product Name'])
d.duplicated().value_counts()
```

```python
# Apply the mapping ONLY to the rows matching those 32 Product IDs
df.loc[df['Product ID'].isin(pro_ids), 'Product Name'] = (df['Product ID'].map(most_frequent_names))
```

```python
# Confirm that one product ID is mapped to one product name
Test3=df.groupby('Product ID')['Product Name'].nunique().sort_values(ascending=False)
Test3.max()
# should be 1
```

**2.6. Feature Engineering - Create new columns**

```python
# Add Year, Month and Day columns from Order Date
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Day'] = df['Order Date'].dt.day

# Create Month_Year column
df['Month_Year'] = df['Order Date'].dt.to_period('M').astype(str)
```

```python
# Add Profit Magin column that calculate Profit per Sales
df['Profit_Margin'] = np.where(df['Sales'] != 0, df['Profit'] / df['Sales'], 0)
```

```python
# Add Shipping Days column that calucate the number of days between order & shippinG
df["Ship Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
```
